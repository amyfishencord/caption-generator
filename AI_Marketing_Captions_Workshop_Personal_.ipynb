```python
# -------------------------------
# Step 3: Paste your API key here
# -------------------------------
import os
from google import genai

# Replace the text Paste API link here with your own API key
os.environ["GOOGLE_API_KEY"] = "PASTE_YOUR_KEY_HERE"

# Create the Gemini client
client = genai.Client(api_key=os.environ["GOOGLE_API_KEY"])
print("API client is ready!")
```

```python
# -------------------------------
# Step 4: Enter product/service to generate captions
# -------------------------------
def generate_captions(product_name):
    prompt = f"Write 3 short, creative, and catchy marketing captions for: {product_name}"
    response = client.models.generate_content(
        model="gemini-2.0-flash",
        contents=prompt
    )
    return response.text

# Ask user for input
product = input("Enter a product or service for marketing captions: ")
captions = generate_captions(product)

print("\nGenerated Marketing Captions:\n")
print(captions)
```
